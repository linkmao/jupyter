## Mcp
Los model context protocol, (MCP) son un aespecie de plugin que usa la IA para usar una herramiena final, pr ejemplo para ue una IA puada acceder a notion, requiere de un MCP que le permita esos alcances, o por ejemplo acceder al calendario de google requiere de un mcp

La configuración de un mcp se hace en el archivo 
`opencode.jsonc`
Ese archivo tiene dos posibles ubicaciones
- A nivel de proyecto (en el directorio raiz del proyecto que se trabaja)
- A nivel global  `~/.config/opencode/opencode.jsonc`

En realidad este es un archivo de configuración general de opencode, lo que significa que no solo se configura el MCP sino tambien muchos otros elementos del agente (temas, shortcuts, etc)

EL siguiente es un ejemplo de como conectar un MCP, para apliar claridad buscar en lod docs de opencode (mcp)

```
{
  "$schema": "https://opencode.ai/config.json",
  "mcp": {
    "jira": {
      "type": "remote",
      "url": "https://jira.example.com/mcp",
      "enabled": true
    
```
La manera de ver los MCP conencatdos es con el comando
/mcp

Algunos mcp que fazt recomienda

- context7 (Se bsuca en la web): Permite a la ia actualizar versioens de paquetes que se usan en el proyecto, como los modelos IA tiene fecha de corte quizas en su entrenamiento no se entera de las nuevas versiones de paquetes y para evitar que busque en internet, este mcp le facilita el asunto

- Chorme dev tools - mcp (se bsuca en github)

## Algunos apuntes importantes y generales de los agentes

Promt
Encuentro que Fazt recomienda un prompmt para iniciar el proyecto inicilmente dese el backend comouna primera etapa, la depura, la testea y luego en una segunda etapa le pide que haga o bien una app web o incluso una de escrirorio, el flujo de trabajo es

- Pedir quehagauna api rest con toda la caracteristicas que se necesita
- Pedir que incluya un CLI para acceder a la api rest via consola
- Luego pedir el forntend

Voy a transcribir como se solicitó los dos primeros elementos anteriores

"Crea una api rest de im sistema administrador de proyectos, el sistema permite registrar proyectos, tarea,as (tablas y kaban), usuarios, equiops de personas, presupuestos, factiras, reunciones, calendario y demas operaciones, esta api lleva autenticacion y autorizacio, rles, testing e2e y unirarios, ademas de usar docker y ofrecer documentación. ademas tambien crea un cli, que usa la api y todo el proyecto es monorepo usando typescript, postgresql y expressjs"

Luego para crear la app de escritorio y de web, donde primeros le pide que haga los planes antes de implementar

"crear dos planes uno para crear una web en vitejs ocn react y una app de escritirio en tauri en donde consumen la api creada"
