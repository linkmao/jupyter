## Rasperry connect
Video de referencia: https://www.youtube.com/watch?v=aYtzX5K3Mf0

Como primer nivel de conexión remota tenemos este servcio que ofrece el mismo rasberry, llamado conect y conect  lite, lo que hace este servicio es **conectarse a la rasperry para gestionarla externamente**, es mas a nivel de administración , ya que difiere a lo otro que busco que es acceder a los contenedores y app instalado en la raspberry a nivel de usuario.

Sin embargo como primer nivel de conexión externa, lo veo muy util
Apreciaciones
connect: se accede remotamente a la interfaz grafica
connect lite: se accede solo al shell

se pueden instalar los dos, en este caso se hará con los dos


### Preliminar
Para que esto funcione la Rasperry debe tener **El inicio de sesion automatico habilitado** es decir, que las rasperry ingrese al sistema sin necesidad de usuario y contraseña.

### Crear cuenta en id rasperry
https://id.raspberrypi.com/   [correo de gmail y contraseña de siempre]
Nota: en el video habaln de connect y de connect-lite, pero actualmente todo se puede hacer con unicamente instalando rpi-connect


### Instalar connect 
Recuerda que se puede instalar de manera independiente
```
sudo apt install rpi-connect
sudo reboot
```
### vinculacion de la cuenta creada con la rasperry
```
rpi-connect signin
```
Genera una url que se debe usar  loguearse y listo
Nota: para que funcione la interfaz grafica debe habilitarse con
```
rpi-connect on
```



## Gestion de doceker desde telegram
A traves de un bot de telegram se puede hacer gestion de los contenedores docker instalados, para ello hay que inatalar un contenedor llamado docker controller bot

Video de referencia: https://www.youtube.com/watch?v=IXERk8heYUc&list=PLi3y-uEs6TANzxOfVtN37tYtd39nZGrQX&index=3

- Crear bot en botfather (entrega un token)
- BUscar el id de usuario telegram (con el bot get_id_bot) mi usuario es 1626793920
- buscar docker controller bot (git gub) y copiar el codigo de docker compose
- crear el archivo docker-compose.yml con el codigo copiado, y editar alli el token y el id de usuario

Crear el contenedor`
docker compose up -`

Listo ya desde telegram se hace gestion de los contenedoresd

```````python
networks:
  wireguard_net:
    driver: bridge
    ipam:
      driver: default
      config:
        - subnet: 10.42.42.0/24
        
services:
  wg-easy:
    image: ghcr.io/wg-easy/wg-easy:15
    container_name: wg-easy
    hostname: wg-easy
    restart: unless-stopped
    networks:
      - wireguard_net
    cap_add:
      - NET_ADMIN
      - SYS_MODULE
    ports:
     # Comprobar puertos libres con comando: sudo ss -tulpn | grep :puerto 
      - "51820:51820/udp"
      - "51821:51821/tcp"
    environment:
      - INSECURE=true
    volumes:
      - ./wireguard:/etc/wireguard # Directorio para archivos persistentes.
      - /lib/modules:/lib/modules:ro
    sysctls:
      - net.ipv4.ip_forward=1
      - net.ipv4.conf.all.src_valid_mark=1
      
  pihole:
    image: pihole/pihole:latest
    container_name: pihole
    hostname: raspberrypi # Nombre de tu dispositivo.
    restart: unless-stopped
    network_mode: host
    environment:
      - TZ=America/Bogota # Zona horaria. Comprobar aquí: https://en.wikipedia.org/wiki/List_of_tz_database_time_zones
      - FTLCONF_webserver_api_password=maolink # Password de acceso a Pi-hole.
     # - FTLCONF_webserver_port=8080,8443s # Opcional: cambiar puertos 80:80 (HTTP) y 443:443 (HTTPS) para acceso web.
      - FTLCONF_dns_listeningMode=ALL
     # - FTLCONF_dns_specialDomains_iCloudPrivateRelay=true # Opcional: relay privado de Apple.
      - PIHOLE_UID=1000 # UID de tu máquina. Comprobar con comando: id -u
      - PIHOLE_GID=1000 # UID de tu máquina. Comprobar con comando: id -g
    volumes:
      - ./pihole:/etc/pihole # Directorio para archivos persistentes.
    cap_add:
     # - NET_ADMIN # Opcional: usar Pi-hole como servidor DHCP.
     # - SYS_TIME # Opcional: usar Pi-hole como cliente NTP.
      - SYS_NICE
 
  duckdns:
    image: lscr.io/linuxserver/duckdns:latest
    container_name: duckdns
    restart: unless-stopped
    network_mode: host
    environment:
      - PUID=1000 # UID de tu máquina. Comprobar con comando: id -u
      - PGID=1000 # UID de tu máquina. Comprobar con comando: id -g
      - TZ=America/Bogota # Zona horaria. Comprobar aquí: https://en.wikipedia.org/wiki/List_of_tz_database_time_zones
      - SUBDOMAINS=maolink # Nombre del subdominio creado en duckdns.org
      - TOKEN=558cb107-f46c-42c9-aa04-39e064f82214 # Token de Duck DNS.
      - UPDATE_IP=ipv4
      - LOG_FILE=false
    volumes:
      - ./duckdns:/config # Directorio para archivos persistentes.
```````

## nginx simple

1. Instalacion de nginx (una unica vez por supuesto)
```
sudo apt install -y nginx
```

2. Dar los permisos a la carpeta donde esta el proyecto
```
sudo chown -R www-data:www-data /carpeta-del-proyecto
sudo chmod -R 755 /carpeta-del-proyecto
```

3. Crear archivo de configuracion del sitio
```
sudo nano /etc/nginx/sites-available/nombre-del-pryecto
```
Este es el texto basico de configuracion

```
server {
    listen 5050;
    listen [::]:5050;

    server_name _;

    direccion-completa-del-proyectomi-web;
    index index.html;

    location / {
        try_files $uri $uri/ =404;

 4. Habilitar el sitio (con un enlace dinamico)

```
sudo ln -s /etc/nginx/sites-available/mi-web /etc/nginx/sites-enabled/
```  5. Verificar que default no este usando el puerto 5050 que es donde se va a desplegar el proyecto ```
sudo cat /etc/nginx/sites-enabled/default | grep listen
```

6. Validar configuracion
```
sudo nginx -t
```

7. Recargar nginx para que tome los cambio
```
sudo systemctl reload nginx
```

8. Abrir el puerto en el firewall (siempre se necesita??? no se)
```
sudo ufw allow 5050/tcp
```

9. Listo
usa desde la consola, para ver el index servido
```
curl http://localhost:5050

o desde la web de cualquier otra maquina local

http://mypi:5050
```

Notas finales:
Si hay problemas de acceso al index.html es porque el usuario www-data necesita permisos de escritura, así que hay que dar esos permisos, acá un ejemplo

```
sudo chmod o+rx /ap
sudo chmod o+rx /ap/hola-mundo
sudo chmod o+r /ap/hola-mundo/index.html
```
