## Busqueda de la ip de rasperry
Despues de que se conecte automaticamente la rasperry en la red, se debe buscar cual es la ip para poder conectarse a ella,
Un comando para powersehll es
`arp -a`
con este comando lista todos los dispositivos conectados en red

## conexion via ssh
```
ssh user@ip
ssh maolink@192.168.78.117
```
En la instalacion del SO en el rasperry esposible ponerle el nombre de la maquina, para que ssh permita esa conecion se debe acceder al archivo **host** de la ruta
`windows/system32/drivers/etc/`
y se agrega la siguiente linea
```
ip-rasperry name
ejemplo
192.168.78.117 mypi.local
```
(al parecer necesita de la palabra local) de esta manera ya la conexion por ssh se puede hacer así:

```
ssh maolink@mypi.local
o incluso me ha servido
ssh maolink@mypi
```

## Instalacion de docker

- Instalar Docker y Docker Compose
sudo curl -fsSL https://get.docker.com/ -o get-docker.sh

sudo sh get-docker.sh

- Añadir nuestro usuario al grupo Docker
sudo usermod -aG docker ${USER}

- Reiniciar sistema
sudo reboot

- Ejecutar contenedor de ejemplo
docker run

```
sudo apt install rpi-connect
sudo reboot
```
### vinculacion de la cuenta creada con la rasperry
```
rpi-connect signin
```
Genera una url que se debe usar  loguearse y listo
Nota: para que funcione la interfaz grafica debe habilitarse con
```
rpi-connect on
```
