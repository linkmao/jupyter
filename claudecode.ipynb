# Claude Code

Incio la documentación de Claude code, el asistente de programación de Athropic y que usa como  modelo Claude, considerado uno de los mas potente. A continuaicón algunas particularidades.

## Instalacion

Instalacion desde POwerShell: 

`irm https://claude.ai/install.ps1 | iex`

Instalacion desde CMD:

`curl -fsSL https://claude.ai/install.cmd -o install.cmd && install.cmd && del install.cmd`

Hay un problema con las variables de entorno (para que se pueda ejecutar claude desde cualquier ruta) la forma de solucionar eso es con los siguientes comandos


`$currentPath = [Environment]::GetEnvironmentVariable('PATH', 'User')``
[Environment]::SetEnvironmentVariable('PATH', "$currentPath;$env:USERPROFILE\.local\bin", 'User'`
)

## Algunos comandos

/init: crea el archivo CLAUDE.MD en el cual hay un resumen tecnico del proyecto que será considerado como contezto
/context: muestra la cantidad de espacio que ocupa el contexto (MCP, skill, mensajes etc)
/clear: limpia el contexto (esto solo se recomienda si etá muy lleno)
/compact: Compacta la conversacion, libera un poco el contexto
/rewind: permite ir una accion atras, es decir te da la opción de borrar la ultima instruccción y el código que generó (es como un CTRL+z)

## Subagentes
Claude tiene la posibilidad de crea y utilizar varios subagentes, cada subagente hace una tarea a la ves, por ejemplo
- Subagente para testear
- Subagente para modificar interfaz
- Subagente para modificar

En ultimas instnciaes esto es para hacer varias tareas a la vez.
