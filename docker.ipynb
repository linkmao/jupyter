# Docker
Nota: con respecto a la id pueden ser los 3 primeros digitos

## Gestion de imagenes
Buscar imagenes

`docker search`

Descargar imagenes

`docker pull [image or id]`

borrar imagen

`docker rm [image or id]`

ver imagenes instalas

`docker images`


## Gestion de contenedores


ejecutar una imagen

`docker run [name image]`

ver las imagenes en ejecucion

`docker ps`

ver todas las contenedores (en ejecucion y no)

`docker ps -a`

ver solamente los id de todas los contenedores
`docker ps -aq`


parar la ejecucion de un contenedor

`docker stop [id or alias]`

reaunadar la ejecucion de un conteneros

`docker start [id or alias]`



Ejecucion en segundo plano
(para que no se quede bloqueda la consola en caso de ser necesario)
docker run -d [image or id]


# Usar el bash de un contenedor
Funciona con ubuntu y con nginx
`docker run -it ubuntu bash`

-it significa iteractivo es decir que el programa queda en ejecucion


# El caso de nginx
Para que el puerto de ngix sea accesible desde el computador local se debe usar la opcion
-p 3000:80, esto hace que se exponga el puesrto 80 donde el contenedor se ejecuta, al puerto 3000 en localhost

`docker run -d -p 3000:80 nginx`


`docker run -d -p [port]:80 --name [my-name] -v [dir origin]:[dir destino]`

ejemplo

`docker run -d -p 3000:80 --name website -v d:/ap/lab/prueba:/usr/share/nginx/html`
nota: la ruta de destino debe ser la mosstrada

Como ingresar al bash de un contenedor en ejecucion
`docker exec -it [name or id] bash`
