## MongoDb

Instalo mongo server, descargandola desde la web
Al momento de instalar de manera completa, debo **deseleccionar** iniciar mongo como servicio (eso haria que siempre mongo se ejecute cuando se inicie el pc, y este no es el caso), pues normalmente en desarrollo inciamos el servicio con 
```
mongod
```

Al instalarse mongo , para la ejecucion de sus programas desde consola se debe llevar la ruta de isntalacion de mongo la variable path del sistema.

La dirección para acceder a la base de datos localmente corresponde a
mongodb://127.0.0.1:27017/bd-name

### Inicio del servicio
Para que mongo funcione debo iniciar el servicio con 
```
mongod
```
**Nota:** Para que pueda montarse el sevidor, se debe crear la ruta c:/data/db

### Uso de shell
```
mongo
```

### Listar base da datos
```
show dbs
```

### Usar o crear una base de datos
Si la base de datos no existe la crea pero no la lista hasta que exista por lo menos una colección (tabla)
```
use [name-db]
```

### Ver base de datos actual
```
db
```

### Borrar una base de datos
```
db.dropDatabase()
```

### Crear una colección
```
db.createCollection("name")
```

### Ver las colecciones existentes en la bd
```
show collections
```

### Borrar una colección
```
db.[name-collection].drop()
```

### Agregar un documento a una coleccion (si la coleccion no existe la crea)
```
db.[name-collection].insert(json)
```

### Agregar multiples datos a una coleccion (se usa un array de json)
```
db.[name-collection].inster([json1, json2,...])
```

### Ver todos los documentos de una coleccion
```
db.[name-collections].find().[pretty()]
```
con pretty() muestra la información indentada

### Buscar un docuemnto en particular 
```
db.[name-collecions].find({prop1:"valor1", prop2:"valor2"})
```

### Buscar el primer elemento que coincida con los parametros dados
```
db.[name-collecions].findOne({prop1:"valor1", prop2:"valor2"})
```

### Filtrar el resultado de la busqueda (para que solo muestre la info que se quiera)
```
db.[name-collecions].find({prop1:"valor1", prop2:"valor2"}, {prop1:1, prop2:1, prop3:0})
```
Aquelllas propiedades que se ponen en 1 se meustran, las que se ponen en cero no se muestran.

### Mostrar una cantidad limite de documentos
```
db.[name-collectios].find({prop:"valor"}).limit(n)
```
n es la cantidad limite a mostrar

### Ver la cantidad total de documentos
```
db.[name-collectios].count()
```

### Actualizar una coleccion
```
db.[name-collection].update({prop:"data1"}, {prop:"data2"})
```
update tiene dos parametros, el primero de ello es un objeto para buscar un documento, y el segundo es el reemplazo que hará, sin embargo es importante saber que update() actualiza toda la coleccion, por lo tanto reemplaza toda la coleccion original.

### Actualizar una coleccion sin perdida de la informacion
```
db.[name-collection].update({prop:"data1"}, {$set:{prop:"data2"}})
```
con el comando $set, se logra adicionar elementos a la coleccion, o modificar un elemento ya existente en la coleccion

### Renombrar el nombre de una propiedad
```
db.[name-coellection].update({prop:"data"},{$rename:{prop:"newname"}})
```
Busca la coleccion y cambia el nomre de la propiedad que se especifica con el comando $rename

### Borrar un documento
```
db.[name-collections].remove({prop:'valor'})
```
Borra todo el documento

### MCP
El Model Context Protocol es un mecanismo que permite a los agentes IA, comunicarse con herramientas externas para realizar alguna tarea, por ejemplo hay MCP que permite conectarse con los calendarios de Google para realizar alguna tarea, o MPC de notion para para que la IA pueda acceder a las funcionalidades de Notion. Los MCP permiten entonces ampliar los "poderes" de la IA pues hace que tenga acciones por fuera del entorno de trabajo.


Los MCP pueden enterdenrse como las "extensiones" de los agentes, algo similar a lo que son las "extensiones" de un IDe como VS-code, solo que en este caso los MCP son usados directamenete por los agentes

A la fecha (abril 2026) los MCP son lo top sin embargo podria haber una posible tendencia a disminuir su uso o desaparecer por la existencia de CLI ya determinados. Esto se explica mas adelante.

En esta documentación dejaré el ejemplo de intaclación de un MCP externo (que no está en el marketplace de VS code) que permite hacer testing a un proyecto, es decir se articula con el agente de AI para planeatr y ejecutar un testting al proyecto el nombre del MCP es *testSprite* (Ver sección mas adelante)


### CLI
Los CLI son programas de consola tipico, como por ejemplo el CLI de Mongo, el CLI de Github, etc. Los CLI sismpre han existido y son formas muy potentes de acceder a todas las funcionalidades de un servicio, ahora, como los agentes pueden ejecutar comandos en los CLI, por eso se dice que los MCP podrian desaparecer pues la IA podria acceder a al conexion de cualquier servicio a traves de su CLI.

### Skill
Este concepto no es mas que instrucciones (prompt) adicionales, especializadas, normalmente ya creadas por empresas o personas especialistas para hacer que el agente haga una tarea.

Por ejemplo hay Skill para que cuando la IA realice el diseño de frontend lo haga bajo unos criterios esteticos definidos. Hay Skill por ejemplo para las buenas practicas en la escritura de código.

Hay un proyecto en github que permite obtener skill
`https://github.com/vercel-labs/skills`

El skill se busca y se instala, por ejemplo este es el enlace para instalar el skill de un ben frontend
npx skills add https://github.com/anthropics/skills --skill frontend-design


los skill se crean en .agents/skill

Los skill son entonces prompt que se inyectan inteligentemente y consecuente al prompt que el usuario le de al agente, por ejemplo si se le pide al agente que use buenas practicas para un diseño de fontend y que use el skill instaldo para ello entonces el agente lo usará


### Agents MD
Esta corresponde a un concepto de instrucción general de como el agente se debe comportar, esto se logra en un archivo llamado AGENTS.md deber colocado en el archivo raiz del proyecto, alli se escribe en lenguaje natural lo que se quiere del comportamiento del agente por ejemplo

- Nombra siempre las carpetas en español
- Nunca uses numeros para la creación del nombre de archivo o carpeta
- usa camelCase para nombrar las carpetas

Bueno y en general lo que se desee del comportamiento del agente. Notese que son cosas sencillas, para cosas mucho mas elaboradas deben usarse skill
